```python
import nltk
```

```python
from nltk.corpus import stopwords
```

```python
def remove_stopwords(text):
    stop_words = set(stopwords.words('english'))
    words = text.split()
    filtered_words = [word for word in words if word.lower() not in stop_words]
    return " ".join(filtered_words)
```

```python
df.head()
```

```python
import nltk
nltk.download('stopwords')
df['overview_no_stopwords'] = df['overview'].apply(remove_stopwords)
print("Stop words removed and stored in 'overview_no_stopwords' column.")
```

```python
from nltk.stem import WordNetLemmatizer
import nltk
nltk.download('wordnet')
print("Imported WordNetLemmatizer and downloaded 'wordnet' corpus.")
```

```python
lemmatizer = WordNetLemmatizer()
def lemmatize_text(text):
    words = text.split() # Simple tokenization by splitting on space
    lemmatized_words = [lemmatizer.lemmatize(word) for word in words]
    return " ".join(lemmatized_words)
df['overview_lemmatized'] = df['overview'].apply(lemmatize_text)
print("Lemmatization applied to 'overview' column and stored in 'overview_lemmatized'.")
```

```python
df.head()
```

```python
from nltk.stem import PorterStemmer
import nltk
nltk.download('punkt')
stemmer = PorterStemmer()
print("PorterStemmer imported and instantiated. 'punkt' tokenizer downloaded.")
```

```python
def stem_text(text):
    words = text.split() # Simple tokenization by splitting on space
    stemmed_words = [stemmer.stem(word) for word in words]
    return " ".join(stemmed_words)
df['overview_stemmed'] = df['overview'].apply(stem_text)
print("Stemming applied to 'overview' column and stored in 'overview_stemmed'.")
```
