```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('/content/IMDB Dataset.csv')
```

```python
missing_values = df[['review', 'sentiment']].isnull().sum()
print("Missing values in 'review' and 'sentiment' columns:\n", missing_values)
```

```python
df['review'] = df['review'].str.lower()
df.head()
```

```python
import re
def remove_html(text):
    clean = re.compile('<.*?>')
    return re.sub(clean, '', text)
df['review'] = df['review'].apply(remove_html)
df.head()
```

```python
def remove_punc_special(text):
    clean = re.sub(r'[^a-zA-Z0-9\s]', '', text)
    return clean
df['review'] = df['review'].apply(remove_punc_special)
df.head()
```

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

# Download NLTK data (if not already downloaded)
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('punkt_tab') # Explicitly downloading punkt_tab as suggested by the error

print("NLTK word_tokenize and stopwords imported, and data downloaded.")
```

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords

# Get the list of English stopwords
stop_words = set(stopwords.words('english'))

def remove_stopwords(text):
    # Tokenize the text
    tokens = word_tokenize(text)
    # Remove stopwords
    filtered_tokens = [word for word in tokens if word not in stop_words]
    # Join the filtered tokens back into a string
    return ' '.join(filtered_tokens)

# Apply the function to the 'review' column
df['review'] = df['review'].apply(remove_stopwords)

# Display the first few rows of the DataFrame to verify the changes
df.head()
```

```python
import nltk
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize

nltk.download('wordnet')

lemmatizer = WordNetLemmatizer()

def lemmatize_text(text):
    tokens = word_tokenize(text)
    lemmatized_tokens = [lemmatizer.lemmatize(word) for word in tokens]
    return ' '.join(lemmatized_tokens)

df['review'] = df['review'].apply(lemmatize_text)

df.head()
```

```python
df['sentiment'] = df['sentiment'].map({'positive': 1, 'negative': 0})
print(df['sentiment'].value_counts())
df.head()
```

```python
all_words = ' '.join(df['review']).split()
total_words = len(all_words)
unique_words = len(set(all_words))

print(f"Total number of words in the corpus: {total_words}")
print(f"Total number of unique words (vocabulary) in the corpus: {unique_words}")
```

```python
df['positive_sentiment'] = df['sentiment'].apply(lambda x: 1 if x == 1 else 0)
df['negative_sentiment'] = df['sentiment'].apply(lambda x: 1 if x == 0 else 0)

df = df.drop('sentiment', axis=1)

print(df[['positive_sentiment', 'negative_sentiment']].head())
df.head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
vectorizer = CountVectorizer(min_df=1, stop_words=None)
```

```python
X = vectorizer.fit_transform(df['review'])
```

```python
vocabulary = vectorizer.get_feature_names_out()
```

```python
import pandas as pd

# Sum the occurrences of each word across all documents (axis=0) directly from the sparse matrix X
total_occurrences = X.sum(axis=0)

# Convert the result to a dense array and then to a pandas Series
word_counts = pd.Series(total_occurrences.tolist()[0], index=vocabulary)

print(f"Vocabulary size: {len(vocabulary)}")
print("\nFirst 10 words in vocabulary:", vocabulary[:10])
print("\nShape of the word counts matrix (number of reviews x vocabulary size):", X.shape)

print("\nTop 20 most frequent words and their total occurrences:")
display(word_counts.sort_values(ascending=False).head(20))
```

### Bag of Bi-grams (n-gram range 1-2)

```python
from sklearn.feature_extraction.text import CountVectorizer

# Initialize CountVectorizer for unigrams and bi-grams
vectorizer_bigram = CountVectorizer(ngram_range=(1, 2), min_df=1, stop_words=None)

# Fit and transform the 'review' column
X_bigram = vectorizer_bigram.fit_transform(df['review'])

# Get the feature names (vocabulary)
vocabulary_bigram = vectorizer_bigram.get_feature_names_out()

print(f"Vocabulary size for unigrams + bi-grams: {len(vocabulary_bigram)}")
print("\nFirst 20 words/bi-grams in vocabulary:", vocabulary_bigram[:20])
print("\nShape of the bi-gram word counts matrix (number of reviews x vocabulary size):", X_bigram.shape)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

# Initialize CountVectorizer for unigrams, bi-grams, and tri-grams
vectorizer_trigram = CountVectorizer(ngram_range=(1, 3), min_df=1, stop_words=None)

# Fit and transform the 'review' column
X_trigram = vectorizer_trigram.fit_transform(df['review'])

# Get the feature names (vocabulary)
vocabulary_trigram = vectorizer_trigram.get_feature_names_out()

print(f"Vocabulary size for unigrams + bi-grams + tri-grams: {len(vocabulary_trigram)}")
print("\nFirst 20 words/n-grams in vocabulary:", vocabulary_trigram[:20])
print("\nShape of the tri-gram word counts matrix (number of reviews x vocabulary size):", X_trigram.shape)
```
